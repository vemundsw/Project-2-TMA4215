The computational time of the numerical approximations are plotted above. The Boris or the RK-4 method are the most efficient depending on the steplength $h_i$. As mentioned previously however, for longer time intervals the Boris or multistep methods would likely give lower global error compared to RK-4, as these methods conserve energy better. In other words, the Boris and RK-4 methods are the most efficient, with lower error for RK-4 for around $t < 3\cdot 10^4$ and Boris for larger $t$.

### Kommentar ang JAX (ikke ta med tittel) (vet ikke helt hvor denne burde plasseres, kanskje før/ rett etter koden for metodene)

We have used the JAX library to cut computational time for all methods. Initially, the methods were implemented with numpy and for loops, but we found this to be an ineffecient approach. In particular, the fourth order symmetric multistep method was too slow. Runs of this method with $T=10^5$ never finished before our patience ran out. We therefore rewrote the code using JAX instead of numpy, as JAX allows for Just In Time compilation (JIT). jax.jit compiles a function into code optimized for your computer (ref 4). The jax.lax.scan function permits JIT compilation on the loops where each iteration of the numerical approximations are computed (ref 5). In short, we used jax.jit and jax.lax.scan to make our code faster.
